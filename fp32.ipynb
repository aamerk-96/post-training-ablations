```python
import os, torch
from transformers import AutoModelForCausalLM
from peft import PeftModel

OUT = os.path.expanduser("~/smol_llm/models/sft-base")
base = AutoModelForCausalLM.from_pretrained("HuggingFaceTB/SmolLM2-360M", dtype=torch.float32)
model = PeftModel.from_pretrained(base, f"{OUT}/adapter")
model.merge_and_unload().save_pretrained(f"{OUT}/merged-fp32")
print(os.listdir(f"{OUT}/merged-fp32"))
```
